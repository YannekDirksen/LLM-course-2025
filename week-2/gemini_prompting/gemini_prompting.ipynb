```python
user_input = """
Generate a high-quality pasta recipe and teach me how to cook it step by step.

Before you answer, think through:
1) What pasta dish is best for beginners but still "chef quality"?
2) What are the failure points?
3) What order of steps prevents those failures?

Then output ONLY:
- Ingredients with quantities (metric)
- Step-by-step method
- Common mistakes to avoid
- Plating + tasting advice
"""
messages.append(user_input)

r = model.generate_content(messages).text
print(r)
```
